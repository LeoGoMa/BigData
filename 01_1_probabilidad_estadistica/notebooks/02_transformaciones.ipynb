```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})
```

```python
df.head
```

```python
df_pais_K = df[df["entity"].str.startswith("K")]
df_pais_K.head(10)
```

```python
df.rename (columns={'entity': 'pais'}, inplace=True)
```

```python
df.set_index('pais')
```

```python
df.loc[df["code"].isnull()]
```

```python
df.loc[df["code"].notnull(),["pais","year","life_expectancy_0"]].head(10)
       
```

```python
df.dropna(subset=["code"], inplace=True)
df.isnull().sum()
```

```python
resultado= (
    df.groupby("pais")["life_expectancy_0"]
    .mean()
    .sort_values(ascending=False)
)
resultado.head(10)
```

```python
df_1970_2023=df.loc[(df["year"]>=1970) & (df["year"]<=2023), ["pais","year","life_expectancy_0"]]
resultado_1970_2023= (
    df_1970_2023.groupby("pais")["life_expectancy_0"]
    .mean()
    .sort_values(ascending=True)
)
resultado_1970_2023.head(10)
```
