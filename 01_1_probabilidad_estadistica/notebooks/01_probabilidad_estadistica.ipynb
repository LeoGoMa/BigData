## Actividad 3 — Estadística descriptiva

**Propósito:** aplicar medidas estadísticas básicas a una variable poblacional.

Se selecciona el año **2020** (periodo histórico) y se filtran los datos correspondientes. Sobre la variable **"Annual change in population"** se calculan: media, mediana, varianza, desviación estándar, mínimo y máximo.

```python
anio = 2020
df_anio = df[df["Year"] == anio]
df_anio.head(3)
```

```python
variable = "Annual change in population"

media = df_anio[variable].mean()
mediana = df_anio[variable].median()
varianza = df_anio[variable].var()
desviacion_estandar = df_anio[variable].std()
minimo = df_anio[variable].min()
maximo = df_anio[variable].max()

print("Media:", media)
print("Mediana:", mediana)
print("Varianza:", varianza)
print("Desviación estándar:", desviacion_estandar)
print("Mínimo:", minimo)
print("Máximo:", maximo)
```

```python
entidad_min = df_anio.loc[df_anio[variable].idxmin(), "Entity"]
entidad_max = df_anio.loc[df_anio[variable].idxmax(), "Entity"]

print("Entidad con el valor mínimo:", entidad_min)
print("Entidad con el valor máximo:", entidad_max)
```

**Preguntas**

1. **¿Cuál es el valor de la media?** ≈ 2,124,908.85 personas.
2. **¿Cuál es el valor de la mediana?** 34,293.5 personas.
3. **¿Cuál es el valor de la desviación estándar?** ≈ 9,522,989.82 personas.
4. **¿Qué entidad presenta el valor mínimo y cuál el máximo?** El valor mínimo lo presenta **Venezuela** (−494,016) y el máximo lo presenta **"World"** (75,707,590). Nota: "World" es una entidad agregada (no un país individual), por lo que representa la suma global, no el país con mayor crecimiento.

## Análisis adicional — País con "A" (2006–2026)

Se filtran los datos de **Argentina** entre los años 2006 y 2026, y se calculan media, mediana, varianza y desviación estándar de "Annual change in population".

```python
pais = "Argentina"
df_pais_fechas = df[(df["Entity"] == pais) & (df["Year"] >= 2006) & (df["Year"] <= 2026)]
df_pais_fechas[["Entity", "Year", "Annual change in population"]]
```

```python
variable_pais = df_pais_fechas["Annual change in population"]

media_pais = variable_pais.mean()
mediana_pais = variable_pais.median()
varianza_pais = variable_pais.var()
desviacion_estandar_pais = variable_pais.std()

print(f"Media ({pais}, 2006-2026):", media_pais)
print(f"Mediana ({pais}, 2006-2026):", mediana_pais)
print(f"Varianza ({pais}, 2006-2026):", varianza_pais)
print(f"Desviación estándar ({pais}, 2006-2026):", desviacion_estandar_pais)
```
